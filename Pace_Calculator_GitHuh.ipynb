```python
from math import trunc
import math
vdot= 0
zones = ["Beginner", "Average", "Decent", "Competitive", "Elite", "Professional"]
zone = ""
def menu():
  print("Running Calculator\n1. Pace\n2. Time\n3. VDOT")
  temp = 0
  while temp not in [1, 2, 3]:
    try:
      temp = int(input(">> "))
    except ValueError:
      print("Enter a correct input")
  if temp == 1:
    pace()
  elif temp == 2:
    time()
  elif temp == 3:
    vdot()

def timeinput():
  count = 0
  while True:
    time = input("Enter a time e.g. [1:05:45]\n>> ")
    for char in time:
      if char == ":":
        count += 1
    if count == 2:
      break
    else:
      print("Invalid format")

def pace():
  while True:
    try:
      unit = input("Distance in Miles or Kilometres? [mi / km]\n>> ").lower()
      if unit in ["mi", "km"]:
        break
    except ValueError:
      print("Invalid format")
  distance = 0
  distance = float(input(f"Enter a Distance in {unit} e.g. [6.34{unit}]\n>>"))
  while distance <= 0:
      distance = float(input(f"Enter a Distance in {unit} e.g. [6.34{unit}]\n>> "))
      print("Invalid format")
  count = 0
  while True:
    time = input("Enter a time e.g. [1:05:45]\n>> ")
    for char in time:
      if char == ":":
        count += 1
    if count == 2:
      break
    else:
      print("Invalid format")

  time = time.split(":")
  min1 = int(time[0]) * 60
  min2 = int(time[2]) / 60
  time = min1 + int(time[1]) + min2
  pace = str(time / distance)
  index = pace.find(".")
  pace1 = pace[:index]
  pace2 = trunc((60 * float((pace[index:]))))
  print(f"You need to run at a {pace1}:{pace2} pace.")


def time():

  distance = 0
  while True:
    try:
      distance = input("Enter distance ran. e.g. [10km]\n>> ")
      if distance[-2:] in ["km", "mi"]:
        break
    except ValueError:
      print("Incorrect format")
  unit = distance[-2:]
  distance = float(distance[:-2])
  while True:
    try:
      pace = input("Enter a pace e.g. [4:15]\n>> ")
      if ":" in pace:
        break
    except ValueError:
      print("Invalid format")
  pace = pace.split(":")
  pace1 = float(pace[0])
  pace2 = float(pace[1]) / 60
  time = (pace1 + pace2) * distance
  seconds = time * 60
  time1 = seconds // 3600
  rem = seconds % 3600
  time2 = rem // 60
  time3 = seconds % 60

  print(f"Running at {int(pace1)}:{pace[1]} for {distance}{unit} will take {int(time1):02d}:{int(time2)}:{int(time3):02d}")

def vdotzones():
  global vdot, zones, zone
  while True:
    try:
      gender = input("Are you male / female?\n>> ").lower()
      if gender in ["male", "female"]:
        break
    except ValueError:
      print("Input a valid gender")
  if gender == "male":
    if vdot < 35:
      zone = zones[0]
    elif vdot < 45:
      zone = zones[1]
    elif vdot < 55:
      zone = zones[2]
    elif vdot < 65:
      zone = zones[3]
    elif vdot < 75:
      zone = zones[4]
    elif vdot > 76:
      zone = zones[5]
  else:
    if vdot < 30:
      zone = zones[0]
    elif vdot < 40:
      zone = zones[1]
    elif vdot < 50:
      zone = zones[2]
    elif vdot < 60:
      zone = zones[3]
    elif vdot < 70:
      zone = zones[4]
    elif vdot > 71:
      zone = zones[5]
def vdot():
  global vdot
  while True:
    try:
      distance = input("Enter distance ran. e.g. [10km]\n>> ")
      if distance[-2:] in ["km", "mi"]:
        break
    except ValueError:
      print("Incorrect format")

  distancem = float(distance[:-2]) * 1000
  count = 0
  while True:
    time = input("Enter a time e.g. [1:05:45]\n>> ")
    for char in time:
      if char == ":":
        count += 1
    if count == 2:
      break
    else:
      print("Invalid format")
  time = time.split(":")
  min1 = int(time[0]) * 60
  min2 = int(time[2]) / 60
  time = min1 + int(time[1]) + min2
  velocity = distancem / time
  vo2 = -4.60 + 0.182258 * velocity + 0.000104 * velocity ** 2
  frac_util = 0.8 + 0.1894393 * math.exp(-0.012778 * time) \
                  + 0.2989558 * math.exp(-0.1932605 * time)
  vdot = vo2 / frac_util
  vdotzones()
  print(f"Your VDOT is: {trunc(vdot)} You're zone is: {zone}")

menu()
```
